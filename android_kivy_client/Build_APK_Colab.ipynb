## 2. Загрузка Android NDK и SDK

````python
import os
import subprocess

# Создаём папку для Android
os.makedirs(os.path.expanduser('~/android'), exist_ok=True)
os.chdir(os.path.expanduser('~/android'))

# NDK
if not os.path.exists('android-ndk-r25b'):
    print('Загружаю Android NDK r25b...')
    os.system('wget -q https://dl.google.com/android/repository/android-ndk-r25b-linux.zip')
    os.system('unzip -q android-ndk-r25b-linux.zip')
    os.system('rm android-ndk-r25b-linux.zip')
    print('✓ NDK загружен')

# SDK
if not os.path.exists('android-sdk'):
    print('Загружаю Android SDK...')
    os.system('wget -q https://dl.google.com/android/repository/commandlinetools-linux-9477386_latest.zip')
    os.makedirs('android-sdk/cmdline-tools', exist_ok=True)
    os.system('unzip -q commandlinetools-linux-*.zip -d android-sdk/cmdline-tools/')
    os.system('mv android-sdk/cmdline-tools/cmdline-tools android-sdk/cmdline-tools/latest')
    os.system('rm commandlinetools-linux-*.zip')
    print('✓ SDK загружен')
    
    print('Установка пакетов Android SDK...')
    os.system('yes | android-sdk/cmdline-tools/latest/bin/sdkmanager "platforms;android-31" "build-tools;33.0.0" > /dev/null 2>&1')
    print('✓ Пакеты установлены')

print('\n✓ Android инструменты готовы')
````

## 3. Клонирование и подготовка проекта

## 4. Сборка APK

````python
import os

# Устанавливаем переменные окружения
os.environ['ANDROID_SDK_ROOT'] = os.path.expanduser('~/android/android-sdk')
os.environ['ANDROID_NDK_HOME'] = os.path.expanduser('~/android/android-ndk-r25b')

os.chdir('/content/OlegShifter/android_kivy_client')

print('Собираю APK...')
print('Это может занять 5-10 минут...')
print('')

result = os.system('buildozer -v android debug')

if result == 0:
    print('')
    print('✓✓✓ APK успешно собран! ✓✓✓')
else:
    print('')
    print('✗ Ошибка при сборке')
````

## 5. Загрузка APK

````python
import os
from google.colab import files

os.chdir('/content/OlegShifter/android_kivy_client')

# Проверяем наличие APK
apk_path = 'bin/olegshifter-1.0.0-debug.apk'

if os.path.exists(apk_path):
    print(f'Загружаю {apk_path}...')
    files.download(apk_path)
    print('✓ APK загружен!')
else:
    print('✗ APK не найден')
    print('Содержимое bin:')
    os.system('ls -la bin/')
````
